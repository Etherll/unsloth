```
REPO = '/content/repo'
JOBS = [{"name": "long_pr_full_A_3408", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "200", "--eval-at", "60,120,200", "--tree", "$ROOT/base", "--seed", "3408", "--out", "$ROOT/out/long_pr_full_A_3408.json"]}, {"name": "long_final_full_A_3408", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "200", "--eval-at", "60,120,200", "--tree", "$ROOT/perf", "--seed", "3408", "--out", "$ROOT/out/long_final_full_A_3408.json"]}, {"name": "long_pr_full_A_3409", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "200", "--eval-at", "60,120,200", "--tree", "$ROOT/base", "--seed", "3409", "--out", "$ROOT/out/long_pr_full_A_3409.json"]}, {"name": "long_final_full_A_3409", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "200", "--eval-at", "60,120,200", "--tree", "$ROOT/perf", "--seed", "3409", "--out", "$ROOT/out/long_final_full_A_3409.json"]}, {"name": "long_pr_full_A_3410", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "200", "--eval-at", "60,120,200", "--tree", "$ROOT/base", "--seed", "3410", "--out", "$ROOT/out/long_pr_full_A_3410.json"]}, {"name": "long_final_full_A_3410", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "200", "--eval-at", "60,120,200", "--tree", "$ROOT/perf", "--seed", "3410", "--out", "$ROOT/out/long_final_full_A_3410.json"]}, {"name": "long400_pr_full_A", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "400", "--eval-at", "60,120,200,300,400", "--tree", "$ROOT/base", "--out", "$ROOT/out/long400_pr_full_A.json"]}, {"name": "long400_final_full_A", "cmd": ["$S/bench.py", "--items", "train", "--warmup", "4", "--mode", "full", "--mb", "2", "--acc", "16", "--steps", "400", "--eval-at", "60,120,200,300,400", "--tree", "$ROOT/perf", "--out", "$ROOT/out/long400_final_full_A.json"]}, {"name": "g1_pr_lora_fp32", "cmd": ["$S/parity.py", "--tree", "$ROOT/base", "--mode", "lora", "--precision", "fp32", "--gc", "true", "--out", "$ROOT/out/g1_pr_lora_fp32.json"]}, {"name": "g1_lstatic_lora_fp32", "cmd": ["$S/parity.py", "--tree", "$ROOT/perf", "--mode", "lora", "--precision", "fp32", "--opts", "lora_compile,trainer_compile", "--gc", "true", "--out", "$ROOT/out/g1_lstatic_lora_fp32.json"]}, {"name": "g1_pr_full_fp32", "cmd": ["$S/parity.py", "--tree", "$ROOT/base", "--mode", "full", "--precision", "fp32", "--gc", "true", "--out", "$ROOT/out/g1_pr_full_fp32.json"]}, {"name": "g1_enccomp_full_fp32", "cmd": ["$S/parity.py", "--tree", "$ROOT/perf", "--mode", "full", "--precision", "fp32", "--gc", "true", "--opts", "enc_compile,trainer_compile", "--out", "$ROOT/out/g1_enccomp_full_fp32.json"]}, {"name": "g1_pr_lora_bf16", "cmd": ["$S/parity.py", "--tree", "$ROOT/base", "--mode", "lora", "--precision", "bf16", "--out", "$ROOT/out/g1_pr_lora_bf16.json"]}, {"name": "g1_lstatic_lora_bf16", "cmd": ["$S/parity.py", "--tree", "$ROOT/perf", "--mode", "lora", "--precision", "bf16", "--opts", "lora_compile,trainer_compile", "--out", "$ROOT/out/g1_lstatic_lora_bf16.json"]}, {"name": "g1_pr_full_bf16", "cmd": ["$S/parity.py", "--tree", "$ROOT/base", "--mode", "full", "--precision", "bf16", "--out", "$ROOT/out/g1_pr_full_bf16.json"]}, {"name": "g1_enccomp_full_bf16", "cmd": ["$S/parity.py", "--tree", "$ROOT/perf", "--mode", "full", "--precision", "bf16", "--gc", "true", "--opts", "enc_compile,trainer_compile", "--out", "$ROOT/out/g1_enccomp_full_bf16.json"]}, {"name": "scr_enccg_full_A", "env": {"PYTHONFAULTHANDLER": "1"}, "print_log": 4000, "cmd": ["$S/bench.py", "--steps", "24", "--warmup", "6", "--mode", "full", "--mb", "2", "--acc", "16", "--tree", "$ROOT/perf", "--opts", "enc_cg", "--out", "$ROOT/out/scr_enccg_full_A.json"]}, {"name": "attn_mb2", "print_log": 3000, "cmd": ["bash", "-c", "[ -d $ROOT/xfpkgs/xformers ] || ($PY -m pip install -q --no-deps --target $ROOT/xfpkgs xformers==0.0.35 && ($PY -m pip install -q --no-deps --target $ROOT/xfpkgs 'fbgemm-gpu-genai==1.6.0' --index-url https://download.pytorch.org/whl/cu130 || echo no-fbgemm)); export PYTHONPATH=$ROOT/xfpkgs; $PY -c 'import torch, xformers; print(\"torch\", torch.__version__, \"xformers\", xformers.__version__)'; $PY -m xformers.info 2>&1 | grep -E 'blackwell|fa2F|cutlassF-pt|pytorch.version'; $PY $S/attn_bench.py --mb 2 --variants sdpa_eff,sdpa_cudnn,varlen_fa2,xf_padded,xf_bd --out $ROOT/out/attn_mb2.json"], "out": "$ROOT/out/attn_mb2.json"}, {"name": "attn_mb8", "print_log": 3000, "cmd": ["bash", "-c", "[ -d $ROOT/xfpkgs/xformers ] || ($PY -m pip install -q --no-deps --target $ROOT/xfpkgs xformers==0.0.35 && ($PY -m pip install -q --no-deps --target $ROOT/xfpkgs 'fbgemm-gpu-genai==1.6.0' --index-url https://download.pytorch.org/whl/cu130 || echo no-fbgemm)); export PYTHONPATH=$ROOT/xfpkgs; $PY -c 'import torch, xformers; print(\"torch\", torch.__version__, \"xformers\", xformers.__version__)'; $PY -m xformers.info 2>&1 | grep -E 'blackwell|fa2F|cutlassF-pt|pytorch.version'; $PY $S/attn_bench.py --mb 8 --variants sdpa_eff,sdpa_cudnn,varlen_fa2,xf_padded,xf_bd --out $ROOT/out/attn_mb8.json"], "out": "$ROOT/out/attn_mb8.json"}, {"name": "c3_eager_full_A0", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 2 --acc 16 --steps 24 --warmup 6 --tree $ROOT/base  --out $ROOT/out/c3_eager_full_A0.json"], "out": "$ROOT/out/c3_eager_full_A0.json"}, {"name": "c3_varlen_full_A0", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 2 --acc 16 --steps 24 --warmup 6 --tree $ROOT/base --opts unpad_varlen --out $ROOT/out/c3_varlen_full_A0.json"], "out": "$ROOT/out/c3_varlen_full_A0.json"}, {"name": "c3_xf_full_A0", "cmd": ["bash", "-c", "export PYTHONPATH=$ROOT/xfpkgs; $PY $S/bench.py --mode full --mb 2 --acc 16 --steps 24 --warmup 6 --tree $ROOT/base --opts unpad_xf --out $ROOT/out/c3_xf_full_A0.json"], "out": "$ROOT/out/c3_xf_full_A0.json"}, {"name": "c3_eager_full_A1", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 2 --acc 16 --steps 24 --warmup 6 --tree $ROOT/base  --out $ROOT/out/c3_eager_full_A1.json"], "out": "$ROOT/out/c3_eager_full_A1.json"}, {"name": "c3_varlen_full_A1", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 2 --acc 16 --steps 24 --warmup 6 --tree $ROOT/base --opts unpad_varlen --out $ROOT/out/c3_varlen_full_A1.json"], "out": "$ROOT/out/c3_varlen_full_A1.json"}, {"name": "c3_xf_full_A1", "cmd": ["bash", "-c", "export PYTHONPATH=$ROOT/xfpkgs; $PY $S/bench.py --mode full --mb 2 --acc 16 --steps 24 --warmup 6 --tree $ROOT/base --opts unpad_xf --out $ROOT/out/c3_xf_full_A1.json"], "out": "$ROOT/out/c3_xf_full_A1.json"}, {"name": "c3_eager_full_B0", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 8 --acc 8 --steps 16 --warmup 6 --tree $ROOT/base  --out $ROOT/out/c3_eager_full_B0.json"], "out": "$ROOT/out/c3_eager_full_B0.json"}, {"name": "c3_varlen_full_B0", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 8 --acc 8 --steps 16 --warmup 6 --tree $ROOT/base --opts unpad_varlen --out $ROOT/out/c3_varlen_full_B0.json"], "out": "$ROOT/out/c3_varlen_full_B0.json"}, {"name": "c3_xf_full_B0", "cmd": ["bash", "-c", "export PYTHONPATH=$ROOT/xfpkgs; $PY $S/bench.py --mode full --mb 8 --acc 8 --steps 16 --warmup 6 --tree $ROOT/base --opts unpad_xf --out $ROOT/out/c3_xf_full_B0.json"], "out": "$ROOT/out/c3_xf_full_B0.json"}, {"name": "c3_eager_full_B1", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 8 --acc 8 --steps 16 --warmup 6 --tree $ROOT/base  --out $ROOT/out/c3_eager_full_B1.json"], "out": "$ROOT/out/c3_eager_full_B1.json"}, {"name": "c3_varlen_full_B1", "cmd": ["bash", "-c", "$PY $S/bench.py --mode full --mb 8 --acc 8 --steps 16 --warmup 6 --tree $ROOT/base --opts unpad_varlen --out $ROOT/out/c3_varlen_full_B1.json"], "out": "$ROOT/out/c3_varlen_full_B1.json"}, {"name": "c3_xf_full_B1", "cmd": ["bash", "-c", "export PYTHONPATH=$ROOT/xfpkgs; $PY $S/bench.py --mode full --mb 8 --acc 8 --steps 16 --warmup 6 --tree $ROOT/base --opts unpad_xf --out $ROOT/out/c3_xf_full_B1.json"], "out": "$ROOT/out/c3_xf_full_B1.json"}]
# Runs on a rented Colab / Kaggle VM (one GPU): builds the base / perf / cand trees from the bench clone,
# installs the pinned stack around the VM's own torch, preps model + data, then runs JOBS in order.
# Every job's JSON comes back on a "LAYA_RESULT <name> <json>" line; progress on "LAYA_JOB" lines.
# Expects REPO (the bench-branch clone) and JOBS (list of {"name", "cmd": [script, args...], "env": {}}).
import json, os, shutil, subprocess, sys, threading, time

BASE_SHA = 'c5cb7cba59f9522d8da2de0522baf77c527f579e'  # placeholder, replaced by cloud_submit.py
PIN = ["transformers==5.5.0", "unsloth_zoo==2026.9.9", "peft==0.21.2", "accelerate==1.15.0", "trl==1.13.0",
       "datasets>=3.4.1,<5", "sentencepiece", "protobuf", "tyro", "hf_transfer", "psutil", "pandas", "pyarrow",
       "pytest", "typer", "rich", "pydantic", "pyyaml", "nest-asyncio", "structlog", "click", "diffusers",
       "sentence-transformers", "safetensors", "huggingface_hub"]
ROOT = "/kaggle/working/lp" if os.path.isdir("/kaggle") else "/content/lp"
S = f"{REPO}/benchmarks/laya_perf/scripts"


def sh(cmd, **kw):
    print("+", cmd, flush = True)
    r = subprocess.run(cmd, shell = True, text = True, capture_output = True, **kw)
    print(r.stdout[-3000:], r.stderr[-3000:], flush = True)
    if r.returncode:
        raise RuntimeError(f"failed: {cmd}")
    return r.stdout


def setup():
    import torch
    os.makedirs(ROOT, exist_ok = True)
    torch_pin = f"torch=={torch.__version__}"
    open(f"{ROOT}/constraints.txt", "w").write(torch_pin + "\n")
    for attempt in range(4):
        try:
            sh(f"{sys.executable} -m pip install -q --timeout 120 -c {ROOT}/constraints.txt " + " ".join(f"'{p}'" for p in PIN))
            break
        except RuntimeError:
            if attempt == 3:
                raise
            time.sleep(20)
    sh(f"git -C {REPO} fetch -q origin {BASE_SHA} laya-decision-perf")
    for name, ref in (("base", BASE_SHA), ("perf", "FETCH_HEAD")):
        if not os.path.isdir(f"{ROOT}/{name}"):
            if name == "perf":
                sh(f"git -C {REPO} fetch -q origin laya-decision-perf")
            sh(f"git -C {REPO} worktree add -q --detach {ROOT}/{name} {ref}")
    if not os.path.isdir(f"{ROOT}/cand"):
        sh(f"git -C {REPO} worktree add -q --detach {ROOT}/cand $(git -C {ROOT}/perf rev-parse HEAD)")
        sh(f"cd {ROOT}/cand && (git apply --check {REPO}/benchmarks/laya_perf/patches/d1_static.diff 2>/dev/null && git apply {REPO}/benchmarks/laya_perf/patches/d1_static.diff && echo applied || echo patch already on perf)")
    for t in ("base", "perf", "cand"):
        print("LAYA_TREE", t, sh(f"git -C {ROOT}/{t} log -1 --format='%H %s'").strip(), flush = True)
    env = os.environ
    env.update(LP_ROOT = ROOT, PY = sys.executable, BASE = f"{ROOT}/base", PERF = f"{ROOT}/perf", CAND = f"{ROOT}/cand", S = S,
               HF_HUB_ENABLE_HF_TRANSFER = "1", TOKENIZERS_PARALLELISM = "false", UNSLOTH_DISABLE_STATISTICS = "1",
               CUDA_VISIBLE_DEVICES = "0", CUDA_DEVICE_ORDER = "PCI_BUS_ID", LANE = "0", PYTHONWARNINGS = "ignore",
               TMPDIR = f"{ROOT}/tmp", AB_OUT = f"{ROOT}/ab")
    os.makedirs(f"{ROOT}/tmp", exist_ok = True)
    os.makedirs(f"{ROOT}/out", exist_ok = True)
    if not os.path.exists(f"{ROOT}/data/items.pt"):
        os.makedirs(f"{ROOT}/data", exist_ok = True)
        from huggingface_hub import snapshot_download
        snapshot_download("convaiinnovations/laya", revision = "55cf4c4ebb", local_dir = f"{ROOT}/model",
                          allow_patterns = ["rl_agent_config.json", "model.safetensors", "encoder/*", "tokenizer/*"])
        sh(f"cd {ROOT} && PYTHONPATH={ROOT}/perf {sys.executable} {S}/prep.py")
    import psutil
    print("LAYA_ENV", json.dumps({
        "torch": torch.__version__, "cuda": torch.version.cuda, "cudnn": torch.backends.cudnn.version(),
        "gpu": torch.cuda.get_device_name(0), "cc": torch.cuda.get_device_capability(0),
        "mem_gib": round(torch.cuda.get_device_properties(0).total_memory / 2**30, 1),
        "bf16": torch.cuda.is_bf16_supported(), "cpus": os.cpu_count(), "ram_gib": round(psutil.virtual_memory().total / 2**30),
        "python": sys.version.split()[0], "pins": sh(f"{sys.executable} -m pip list 2>/dev/null | grep -iE '^(transformers|unsloth.zoo|peft|accelerate|triton|trl|xformers|flash.attn) '"),
        "cpu_model": open("/proc/cpuinfo").read().split("model name")[1].split("\n")[0].strip(": ") if os.path.exists("/proc/cpuinfo") else "?",
    }), flush = True)


def heartbeat(name, stop):
    t0 = time.time()
    while not stop.wait(240):
        print(f"LAYA_JOB {name} running {int(time.time() - t0)} s", flush = True)


def run(job):
    name, cmd, env = job["name"], job["cmd"], {**os.environ, **{k: str(v) for k, v in job.get("env", {}).items()}}
    cmd = [x.replace("$ROOT", ROOT).replace("$S", S).replace("$PY", sys.executable) for x in cmd]
    if cmd[0].endswith(".py"):
        cmd = [sys.executable] + cmd
    out = job.get("out", "").replace("$ROOT", ROOT) or next((cmd[i + 1] for i, x in enumerate(cmd) if x == "--out"), None)
    log = f"{ROOT}/out/{name}.log"
    stop = threading.Event()
    threading.Thread(target = heartbeat, args = (name, stop), daemon = True).start()
    t0 = time.time()
    with open(log, "w") as f:
        rc = subprocess.run(cmd, stdout = f, stderr = subprocess.STDOUT, env = env, cwd = job.get("cwd", ROOT).replace("$ROOT", ROOT)).returncode
    stop.set()
    print(f"LAYA_JOB {name} rc={rc} {time.time() - t0:.0f}s", flush = True)
    if job.get("print_log"):
        print(f"LAYA_LOG {name} " + json.dumps(open(log).read()[-job["print_log"]:]), flush = True)
    if rc:
        print(f"LAYA_FAIL {name} " + json.dumps(open(log).read()[-4000:]), flush = True)
    elif out and os.path.exists(out):
        print(f"LAYA_RESULT {name} " + open(out).read().replace("\n", " "), flush = True)
    return rc


setup()
failed = [j["name"] for j in JOBS if run(j)]
print("LAYA_DONE", json.dumps({"jobs": len(JOBS), "failed": failed}), flush = True)
```
